**Break: It is used to completely stop the loop when a specific condition is met.**

```python
#while loop
while True:
  print("Type your name")
  name=input(">")
  if name=="your name":
    break
  print("Thank You!")
```

**Continue: It is used to skip the current iteration and move to the next iteration of the loop.**

```python
for num in range(10):
  if num%2==0:
    continue
  print(num)
```

```python
# This is a guess the number game.
import random

secret_number = random.randint(1, 20)

print('I am thinking of a number between 1 and 20.')

# Ask the player to guess 6 times.
for guesses_taken in range(1, 7):
    print('Take a guess.')
    guess = int(input('>'))
    if guess < secret_number:
        print('Your guess is too low.')
    elif guess > secret_number:
        print('Your guess is too high.')
    else:
        break # This condition is the correct guess!
if guess == secret_number:
    print('Good job! You got it in ' + str(guesses_taken) + ' guesses!')
else:
    print('Nope. The number was ' + str(secret_number))





```
