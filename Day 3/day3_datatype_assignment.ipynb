```python
# Numbers

x = -25
y = 4.75

print(abs(x))
print(round(y))
print(pow(2, 3))
print(divmod(17, 5))
print(int(y))
```

```python
# String

text = "Hello World"

print(text.upper())
print(text.lower())
print(text.split())
print(text.replace("Hello", "world"))
print(text.find("Hello"))
print(len(text))
```

```python
# List

numbers = [10, 20, 30, 20]

numbers.append(40)
print(numbers)

numbers.insert(1, 15)
print(numbers)

numbers.remove(20)
print(numbers)

print(numbers.pop())

numbers.sort()
print(numbers)

print(numbers.count(20))
```

```python
# Tuple

items = ("apple", "banana", "apple", "mango")

print(items.count("apple"))
print(items.index("banana"))
print(len(items))
print(sorted(items))
print(tuple(["x", "y", "z"]))
```

```python
# Dictionary

student = {
    "name": "Tanishka",
    "age": 20,
    "course": "AIML"
}

print(student.keys())
print(student.values())
print(student.items())
print(student.get("name"))

student.update({"city": "Nashik"})
print(student)

print(student.pop("age"))
```

```python
# Set

num = {10, 20, 30, 40}
a = {30, 40, 50, 60}

num.add(50)
print(num)

num.remove(20)
print(num)

print(num.union(a))
print(num.intersection(a))
print(num.difference(a))
print(len(num))
```

```python
# Boolean

a = True
b = False

print(bool(1))
print(bool(0))
print(bool("Hii"))
print(bool(""))
print(int(a))
print(str(b))
```

```python
# None

value = None

print(type(value))
print(bool(value))
print(str(value))
print(value is None)
print(value == None)
```
