```python
import numpy as np

# Defining Activation Function Sigmoid
def sig(x):
    return 1 / (1 + np.exp(-x))

#Defining inputs, weights and bias
inputs = np.array([1, 2])
weights = np.array([0.5, 0.3])
bias = 0.1

# calculating the weighted sum
sum = np.dot(weights, inputs) + bias

# Applying Activation Function
r = sig(sum)

print("Neuron Output:", r)
```

In this we used dot product to calculate the weighted sum for that the formula is y=xi * wi + bias

and then we applyed activation function sigmoid for that the formula is 1/1+e^-x

the bias is applied here to add some extra value to the input

###Question 2

```python
import torch
import torch.nn as nn
# Define the input data
x = torch.tensor([[0.1, 0.2]])
#Shape
print('Input:', x.shape)

linear_layer = nn.Linear(in_features=2,
                         out_features=1,
                         bias = True,
                         dtype= torch.float)
# Apply the linear transformation to the input data
y = linear_layer(x)

#print the outputs
print('output:',y)

print("PyTorch Weights:", linear_layer.weight.data)
print("PyTorch Bias:", linear_layer.bias.data)
```

Here in the above diagram the flow of How Large language models predicts the tokens is shown


**Tokenization**: so first the tokenization is done and each word has given a seperate numerical representation i.e token number.


**Embeddings**:Then the tokens are stored as Embeddings in a vector database in a matrix format with their dimensions.


**Transformer Layer**:Then the transformer layer helps to understand the meanings of the embeddings


**Next Token Prediction**:Then based on the highest probability of the token the next token is generated mostly the softmax function is used to calculate the probabilities

**What is Predicting a Next token in LLMs?**

Predicting a next token means guessing the most relevant word or part of a word or a token that should come next.

A large language model predicts the next token by converting text into numerical values, processing them and calculate probabilities for every words in a database and according to the rankings it selects the top most token and generates it.
