```python
# Queation 1

# Our list of numbers
numbers = [12, 7, 9, 4, 22, 15, 8, 3, 30, 11]

# Start both counters from zero
even_count = 0
odd_count = 0

# Defining an empty list
even=[]

# Go through each number one by one
for i in numbers:
    # Check if the number can be divided by 2
    if i % 2 == 0:
        print(i, "is even")
        even_count += 1
        even.append(i)
    else:
        # If it is not even, then it is odd
        print(i, "is odd")
        odd_count += 1

# Show the final counts and the list of even numbers
print("Count Even Numbers:", even_count)
print("Count Odd Numbers:", odd_count)
print("Even Numbers:", even)
```

```python
# Question 2

'''Function to calculate the average of a list of numbers'''
def average(numbers: list) -> float:

    ''' Take the sum of all numbers and divide it by the length of the list'''
    return sum(numbers) / len(numbers)


''' Create an empty list to store the numbers entered by the user'''
numbers = [10,20,30]


try:
    # Calculate and display the average
    print("Average:", average(numbers))

except ZeroDivisionError:
    # This message is shown if the list is empty
    print("Can't calculate the average!!")
```

```python
# Question 3

import random

numbers = []

# Generate 10 random guesses
for i in range(1, 11):
    guess = random.randint(1, 50)
    numbers.append(guess)

    # Print each guess as it is generated
    print(f"Guess {i}: {guess}")

# Find the highest guess
highest = max(numbers)

# Calculate the average of all guesses
average = sum(numbers) / len(numbers)

# Print the final results
print("All guesses:", numbers)
print("Highest guess:", highest)
print("Average guess:", round(average, 2))
```

```python
# Question 4

class Student:
    ''' Set up the student's name and marks when the object is created'''
    def __init__(self, name: str, marks: list):
        self.name = name
        self.marks = marks

    ''' Calculate the average of the student's marks'''
    def average(self) -> float:
        return sum(self.marks) / len(self.marks)

    ''' Decide how the student object should look when we print it'''
    def __str__(self) -> str:
        return f"{self.name}: average {self.average()}"


# Create three Student objects with their names and marks
stud1 = Student("John", [10, 20, 30])
stud2 = Student("David", [40, 50, 60])
stud3 = Student("Lily", [70, 80, 90])

# Put all the students into one list
students = [stud1, stud2, stud3]

# Start with zero as the highest average
high = 0
h_stud = ""

# Go through each student one by one
for i in students:
    # Print the student's name and average
    print(i)

    # Check if this student's average is higher than the current highest
    if i.average() > high:
        high = i.average()
        h_stud = i.name


# Display the name of the student with the highest average
print("Student with highest average:", h_stud)

# Display the highest average
print("Highest average:", high)
```
