```python
# Program to demonstrate ZeroDivisionError

def zerodiv(a,b):
  try:
    return a / b
  except ZeroDivisionError:
    print("Error: Cannot divide by zero!")
num1=int(input("Enter a number:"))
num2=int(input("Enter a number:"))
print(zerodiv(num1,num2))

```

```python
# Program to demonstrate ValueError
def valerror(s):
  try:
  # Attempt to convert a string to an integer
    num = int(s)
    return num
  except ValueError:
    print("Error: Could not convert string to an integer. Invalid input value")
string=str(input("Enter a String:"))
print(valerror(string))
```

```python
# Program to demonstrate TypeError

def tyerror(n,s):
  try:
    # Attempt to add an integer and a string
    return n + s
  except TypeError:
    print("Error: Cannot perform addition between an integer and a string.")

num=int(input("Enter a number:"))
string=str(input("Enter a String:"))
print(tyerror(int,str))
```

```python
# Program to demonstrate IndexError
def inderror(n):
  try:
     l = [1, 2, 3]
    # Attempt to access an index that does not exist
     i = l[n]
     print(f"Element at index {n}:{i}")
  except IndexError:
     print("Error: List index out of range. The index you tried to access does not exist.")
num=int(input("Enter a list index:"))
print(inderror(num))
```
