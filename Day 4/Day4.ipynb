try block: code that could potentially fail goes here. python attempts to run it normally

except block : if an error happens inside try execution jumps here immediately instead of crashing

```python
try:
  42/0
except ZeroDivisionError:
  print("Error: Division by zero")
```

```python
def divide_by(n):
  return 42/n

print(divide_by(2))
print(divide_by(12))
print(divide_by(0))
print(divide_by(4))


```

```python
def spam(divided_by):
  try:
    return 42/divided_by
  except ZeroDivisionError:
    print("Error: Invalid Argument")

print(spam(2))
print(spam(12))
print(spam(0))
print(spam(4))
```

```python
#this is animation program
import time, sys
indent = 0 # How many spaces to indent
indent_increasing = True # Whether the indentation is increasing or not
try:
    while True: # The main program loop
        print(' ' * indent, end='')
        print('********')
        time.sleep(0.1) # Pause for 1/10th of a second.
        if indent_increasing:
            # Increase the number of spaces:
            indent = indent + 1
    if indent == 20:
        # Change direction:
        indent_increasing = False
    else:
        # Decrease the number of spaces:
        indent = indent - 1
    if indent == 0:
        # Change direction:
        indent_increasing = True
except KeyboardInterrupt:
    sys.exit()
```
