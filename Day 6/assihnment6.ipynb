```python
# Decorator using decorator syntax
def decorator(function):

    def wrapper():
        print("Hello my name is Tanishka")
        function()

    return wrapper

@decorator
def say_hello():
    user_name = str(input("enter your name"))
    print(f"hey {user_name}")

say_hello()
```

**Context Managers**

```python
# Without Context Manager
file = open('data.txt',"w")
content = file.write("Good Morning, Have a great day")
file.close()
```

```python
file = open('data.txt',"r")
content = file.read()
print(content)
file.close()
```

```python
# With Context Manager
with open('data.txt',"w") as file:
    content = file.write("Good Morning, Have a great day")

with open('data.txt',"r") as file:
    content = file.read()
    print(content)
```

But with context managers we don't need to close the file explicitly python automatically closes the file after the block is executed or the error is occurred

```python
class MyContext:
    def __enter__(self):
        print('Entering')
        return self

    def __exit__(self, exc_type, exc_value, tb):
        print('Leaving')

with MyContext():
    print('Inside')
```

```python
class MyContext:
    def __enter__(self):
        print("Entering")
        return self

    def __exit__(self, exc_type, exc_value, tb):
        print("Leaving")
        print("Exception Type:", exc_type)
        print("Exception Value:", exc_value)
        print("Traceback:", tb)


with MyContext():
    print("Inside")
    l=[1,2,3]
    print(l[3])
```

```python
import time #Used to acess time
from functools import wraps


def show_time(function): #Decorator

    @wraps(function)
    def wrapper(*args, **kwargs):
        start = time.time() #records the time before the function is started
        result = function(*args, **kwargs)
        print(f"Time: {time.time() - start:.4f}s") # time required to execute the function is calculated
        return result

    return wrapper


@show_time #decorator
def train_one_epoch(): # Function to be timed
    for _ in range(10):
        time.sleep(1)
```

```python
#micro project

import time
from contextlib import contextmanager # Importing contextmanager to create a custom context manager using a function

@contextmanager
def timer():
    # Marking the starting time
    start = time.time()

    try:
        yield

    finally:
        print(f"Time: {time.time() - start:.4f}s") # The execution time required by the function will be printed

def train_one_epoch():
    time.sleep(5) # making it to sleep for 5 seconds


with timer():
    train_one_epoch()
```
