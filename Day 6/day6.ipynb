```python
class Countdown:
  def __init__(self, start):
    self.current = start # current is initializes with start

  def __iter__(self):
    return self

  def __next__(self):
    if self.current < 0:

      # exception is raised if the value of current goes less than 0
       raise StopIteration

    number=self.current # number stores current number before decrementing it.
    self.current-=1 # current value is decremented by 1
    return number

for number in Countdown(3): # calls iter and next method repeatedly
  print(number)

```

```python
def count_up_to(limit):
    number = 1
    while number <= limit:
        yield number
        number += 1
for number in count_up_to(5): # Before the loop starts the function has all the values for the loop and prints it as the iteration starts
    print(number)
```

```python
# Creating a list from range(1, 1_000_001) will store all 1,000,000 numbers in memory at once.
# This can consume large amount of memory
numbers = list(
    range(1, 1_000_001)
)
# all 1,000,000 in memory
```

```python
def numbers():
    # This is a generator function. It produces numbers one at a time
    for n in range(1, 1_000_001):
        yield n
number=numbers()
print(next(number))
print(next(number))
print(next(number))
```

```python
# Decorator by a lengthy way
def decorator(function): # Passing a baseline function to the decorator function as a parameter

    def wrapper():
        print("Hello my name is Tanishka")
        function()
    return wrapper # Returns wrapper function to the decorator function

# The function to be decorated.
def say_hello():
    user_name = str(input("enter your name"))
    print(f"hey {user_name}")

say_hello = decorator(say_hello)
say_hello()
```
